```python
import pandas as pd
```

```python
df = pd.read_csv('Campionato.csv', sep=';')
```

```python
df.head()
```

```python
squadra = 'Fiorentina'
df = df[['Home', 'Away', 'Result']]
df.query(f'Home == "{squadra}" | Away == "{squadra}"').head(8)
```

```python
risultati = []
risultato = '1'  # Example: '1' for home win, '2' for away win, 'X' for draw
matches = df.query(f'Home == "{squadra}" or Away == "{squadra}"')
precedente = 0
for _, row in matches.iterrows():
    risultato = str(row['Result']).strip()
    if risultato == 'X':
        new = precedente
    else:
        if row['Home'] == squadra:
            if risultato == '1':
                new = precedente + 1
            elif risultato == '2':
                new = precedente - 1
            else:
                new = precedente
        else:  # Juventus is away
            if risultato == '2':
                new = precedente + 1
            elif risultato == '1':
                new = precedente - 1
            else:
                new = precedente
    risultati.append(new)
    precedente = new
print(risultati)
```

```python

esiti = []

squadraInEsame = df.query(f'Home == "{squadra}" or Away == "{squadra}"')

for _, row in squadraInEsame.iterrows():    #itero la series
    esito = str(row['Result']).strip()  #estraggo l'esito 
    if esito == 'X': # se l'esito è X in casa o trasferta non cambia
        new = 'X'
    else:
        if row['Home'] == squadra:
            if esito == '1':    # se gioca in casa 1 rimane 1 ma va aggiunto alla lista
                new ='1'
            elif esito == '2':  # se gioca in casa e perde resta 2 e va aggiunto alla lista
                new = '2'
            
        else:  # squadra is away
            if esito == '2': # se gioca in trasferta e vince diventa 1
                new = '1'
            elif esito == '1':  # se gioca in trasferta e perde diventa 2
                new = '2'
            
    esiti.append(new)
print(esiti)
```
